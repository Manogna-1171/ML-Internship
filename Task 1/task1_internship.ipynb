## **2. Identified Data Types**
###Student Performance Dataset – Columns

**StudentID** – Numerical – Unique ID for each student

**Age** – Numerical (Discrete) – Age of the student in years

**Gender** – Categorical (Binary) – 0 = Female, 1 = Male

**Ethnicity** – Categorical (Nominal) – Encoded ethnicity group of the student

**ParentalEducation** – Categorical (Ordinal) – Education level of parents (ordered from low to high)

**StudyTimeWeekly** – Numerical (Continuous) – Weekly study time in hours

**Absences** – Numerical (Discrete) – Number of days the student was absent

**Tutoring** – Categorical (Binary) – 0 = No, 1 = Yes

**ParentalSupport** – Categorical (Ordinal) – Level of parental support (low → high)

**Extracurricular** – Categorical (Binary) – 0 = No, 1 = Yes

**Sports** – Categorical (Binary) – 0 = No, 1 = Yes

**Music** – Categorical (Binary) – 0 = No, 1 = Yes

**Volunteering** – Categorical (Binary) – 0 = No, 1 = Yes

**GPA** – Numerical (Continuous) – Grade Point Average of the student

**GradeClass** – Categorical (Ordinal) – Final grade category of the student (TARGET VARIABLE)

## **5. Identifying Target Variable and Input Features**

- The target variable is the column that the model aims to predict.

- In the Student Performance dataset, GradeClass is the target variable.

- It represents an ordinal multi-class classification problem (performance categories ordered from low to high).

- Input features are the columns used to make predictions.

- Features such as Age, Gender, Ethnicity, ParentalEducation, StudyTimeWeekly, Absences, Tutoring, ParentalSupport, Extracurricular, Sports, Music, Volunteering, and GPA are suitable for modeling.

- Identifier-like columns such as StudentID are not useful for prediction and should be removed before training.

```python
#If we want to predict the survival based on the dataset we can deop 'PassengerId' and 'ticket'
train_dataset_filter = train_dataset.drop(columns=['StudentID']) # Once run,we cannot take back the data if put into same variable
```
