```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv('Titanic-Dataset.csv')
```

```python
df.head()
```

#Reviewing the important features

```python
# review important features like Age, Sex, Fare, and Survived.
df.Age.isnull().sum()
```

```python
df.Sex.isnull().sum()
```

```python
df.Fare.isnull().sum()
```

```python
df.Survived.unique().sum()
```

```python
df['Survived'].value_counts()# these values are un evenly distributed
```

```python
df_filtered=df.copy()
```

#Replacing the null values with median and mode

```python
df_filtered['Age'].describe().loc['50%']
```

```python
df_filtered['Age']=df_filtered['Age'].fillna(df_filtered['Age'].describe().loc['50%'])
```

```python
df_filtered['Embarked']=df_filtered['Embarked'].fillna(df_filtered['Embarked'].mode()[0])
```

```python
df_filtered.isnull().sum()
```

#  Removing columns that are not nessasary

```python
df_filtered=df_filtered.drop(columns=['PassengerId','Name'],axis=0)
```

```python
df_filtered.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe=OneHotEncoder(handle_unknown='ignore',sparse_output=False).set_output(transform="pandas")
encoded=ohe.fit_transform(df_filtered[['Sex','Embarked']])
```

```python
encoded
```

```python
df_filtered=df_filtered.drop(columns=['Sex','Embarked'])
```

```python
df_filtered['Embarked_C']=encoded['Embarked_C']
df_filtered['Embarked_Q']=encoded['Embarked_Q']
df_filtered['Embarked_S']=encoded['Embarked_S']
df_filtered
```

```python
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
```

```python
sc=StandardScaler().set_output(transform="pandas")
after_scaling=sc.fit_transform(df_filtered[['Age','Fare']])
```

```python
after_scaling
```

```python
df_filtered['age_scaled']=after_scaling['Age']
df_filtered['fare_scaled']=after_scaling['Fare']
df_filtered=df_filtered.drop(columns=['Age','Fare'])
```

```python
df_filtered
```

```python
x=df_filtered.drop(columns=['Survived'])
y=df_filtered['Survived']
x
```

```python
x=x.drop(columns=['Cabin','Ticket'])
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=40)
```

```python
x_train.shape
x_test.shape
```

```python
model=LogisticRegression()
model.fit(x_train,y_train)
```

```python
y_predict=model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,precision_score,f1_score,recall_score,confusion_matrix
```

```python
accuracy=accuracy_score(y_test,y_predict)
accuracy
```

```python
precision=precision_score(y_test,y_predict)

recall=recall_score(y_test,y_predict)
f1=f1_score(y_test,y_predict)

```

```python
print(accuracy)
print(f1)
print(precision)
print(recall)
```

```python
confusion_matrix=confusion_matrix(y_test,y_predict)
confusion_matrix
```

```python
from sklearn.metrics import roc_curve, auc
y_prob = model.predict_proba(x_test)[:, 1]

# Calculate ROC values
fpr, tpr, thresholds = roc_curve(y_test, y_prob)

# Calculate AUC

roc_auc = auc(fpr, tpr)
```
