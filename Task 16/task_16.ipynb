# **hypertuning using grid search cv**

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
# from sklearn.preprocessing import Logi
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

## loading the breastcancer dataset

```python
data=load_breast_cancer(as_frame=True)
df=data.frame
x=df.drop(columns="target")
y=df["target"]
```

# preprocessing the dataset

```python
df.isnull().sum()
```

```python
numerical_features=x.select_dtypes(include=['float64']).columns
catogirical_features=x.select_dtypes(include=['int64']).columns
```

```python
ss=StandardScaler().set_output(transform="pandas")
x_scaled=ss.fit_transform(x)
x_scaled
```

# spiting the dataset to training and testing

```python
x_train,x_test,y_train,y_test=train_test_split(x_scaled,y,test_size=0.2,random_state=42)
```

# applying grid_searchcv

```python
clf=GridSearchCV(SVC(gamma="auto"),{
    "C":[1,10,20],
    "kernel":["rbf","linear"]
},cv=5,return_train_score=False)
clf.fit(x_train,y_train)
df2=pd.DataFrame(clf.cv_results_)
df2
```

```python
df2[['params','mean_test_score']]
```

```python
clf.best_score_
```

```python
clf.best_params_
```

# comparing the tuned model with the default model

```python
best_model=clf.best_estimator_
y_pred=best_model.predict(x_test)
tuned_accuracy=accuracy_score(y_test,y_pred)
print("classificationreport of best model" ,classification_report(y_test,y_pred))
```

```python
ss=SVC(kernel="linear",C=10)
ss.fit(x_train,y_train)
y_pred=ss.predict(x_test)
default_accuracy=accuracy_score(y_test,y_pred)
print("classification_report of default_model is",classification_report(y_test,y_pred))
print("accuracy difference is ",(tuned_accuracy-default_accuracy)*100)
```
